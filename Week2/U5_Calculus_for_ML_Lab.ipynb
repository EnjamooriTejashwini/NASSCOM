```python

def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
import sympy as sp

def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 (use h = 1e-6)
x = 2
h = 1e-6

numerical_derivative = (g(x + h) - g(x)) / h

print("1. Numerical derivative at x = 2:")
print(numerical_derivative)

# 2. Symbolic derivative of x**3 + 2*x
X = sp.symbols('x')
expression = X**3 + 2*X

symbolic_derivative = sp.diff(expression, X)

print("\n2. Symbolic derivative:")
print(symbolic_derivative)

# 3. Evaluate symbolic derivative at x = 2 and compare
symbolic_value = symbolic_derivative.subs(X, 2)

print("\n3. Symbolic derivative at x = 2:")
print(symbolic_value)

print("\nComparison:")
print("Numerical derivative =", numerical_derivative)
print("Symbolic derivative  =", symbolic_value)
```

```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')

# Define function
f2 = x**2 + 3*x*y + y**2

# Partial derivatives
print('f(x, y) =', f2)

print('df/dx =', sp.diff(f2, x))   # 2*x + 3*y
print('df/dy =', sp.diff(f2, y))   # 3*x + 2*y
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)

# Evaluate the gradient at the point (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)   # points in the steepest-ascent direction
```

```python
import sympy as sp

# Define symbols
x, y = sp.symbols('x y')

# Define function
h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

# 2. Assemble the gradient list
gradient = [dh_dx, dh_dy]

print("\nGradient =", gradient)

# 3. Evaluate at (x=2, y=0)
gradient_at_point = [g.subs({x: 2, y: 0}) for g in gradient]

print("\nGradient at (x=2, y=0) =", gradient_at_point)
```

```python
import sympy as sp

# Define the variable
x = sp.symbols('x')

# Define y = exp(x**2 + 1)
y = sp.exp(x**2 + 1)

# 1. By hand (Chain Rule)
# Outer derivative = exp(x**2 + 1)
# Inner derivative = 2*x
dy_dx_hand = sp.exp(x**2 + 1) * (2*x)

print("1. By hand:")
print(dy_dx_hand)

# 2. With sp.diff
dy_dx_sympy = sp.diff(y, x)

print("\n2. Using sp.diff:")
print(dy_dx_sympy)

# 3. Confirm they match
print("\n3. Do they match?")
print(sp.simplify(dy_dx_hand - dy_dx_sympy) == 0)
```

```python
# Tiny toy problem: 4 samples, 3 input features, 5 hidden units, 1 output
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')

```

```python
lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1     # hidden -> output

# ReLU function
def relu(x):
    return np.maximum(0, x)

# 1. Forward pass
z1 = Xb @ Wa
h = relu(z1)
y_hat = h @ Wb

# Mean Squared Error loss
loss_before = np.mean((y_hat - Yb) ** 2)

print("Loss before update:", loss_before)

# 2. Backward pass
n = Xb.shape[0]

# Derivative of MSE loss
dy = (2 / n) * (y_hat - Yb)

# Gradient for Wb
dWb = h.T @ dy

# Gradient for hidden layer
dh = dy @ Wb.T

# Gradient through ReLU
dz1 = dh * (z1 > 0)

# Gradient for Wa
dWa = Xb.T @ dz1

# 3. One step with learning rate = 0.05
lr = 0.05

Wa = Wa - lr * dWa
Wb = Wb - lr * dWb

# Forward pass again after update
z1_new = Xb @ Wa
h_new = relu(z1_new)
y_hat_new = h_new @ Wb

loss_after = np.mean((y_hat_new - Yb) ** 2)

print("Loss after update:", loss_after)
```

```python
import sympy as sp

# Define variables
x, y = sp.symbols('x y')

# 1. Hessian of x**4 + y**2
f = x**4 + y**2

H = sp.hessian(f, (x, y))

print("1. Hessian:")
sp.pprint(H)

# 2. Gradient descent to minimise (x - 7)**2
xv = 0.0
lr = 0.1

for step in range(20):
    # Derivative of (x - 7)^2
    grad = 2 * (xv - 7)

    # Gradient descent update
    xv = xv - lr * grad

print("\n2. Final x:", xv)
```
